# Energy Forecasting Experiments

This notebook contains experiments and trials for the energy forecasting pipeline.

```python
# Import necessary libraries
import pandas as pd
import numpy as np
from pathlib import Path

# Set paths
DATA_PATH = "/home/moukouba/equilibrium/model_ready.parquet"
OUTPUT_DIR = Path("outputs")
OUTPUT_DIR.mkdir(exist_ok=True)

print(f"Data path: {DATA_PATH}")
print(f"Output directory: {OUTPUT_DIR}")
```

```python
# Load data
df = pd.read_parquet(DATA_PATH, engine='fastparquet')
print(f"Loaded {len(df)} rows")
print(f"Columns: {len(df.columns)}")
df.head()
```

```python
# Target columns
TARGET_COLS = [
    "da_energy_aeci_lmpexpost_ac",
    "da_energy_michigan_hub_lmpexpost_ac",
    "da_energy_minn_hub_lmpexpost_ac",
]

print("Target columns:")
for col in TARGET_COLS:
    print(f"  - {col}")
```

```python
# Basic statistics
print("\nBasic Statistics:")
for col in TARGET_COLS:
    print(f"\n{col}:")
    print(f"  Mean: {df[col].mean():.2f}")
    print(f"  Std: {df[col].std():.2f}")
    print(f"  Min: {df[col].min():.2f}")
    print(f"  Max: {df[col].max():.2f}")
```

```python
# Train/validation split
split_idx = int(0.85 * len(df))
train_df = df.iloc[:split_idx]
val_df = df.iloc[split_idx:]

print(f"Training samples: {len(train_df)}")
print(f"Validation samples: {len(val_df)}")
```

```python
# Feature engineering trial
print("\nFeature Engineering Trials:")
print("  - Lags: [1, 2, 3, 6, 12, 24]")
print("  - Rolling windows: [3, 6, 12, 24]")
print("  - EWM spans: [12, 24]")
print("  - Differencing: [1, 24]")
```

```python
# Model configuration trials
print("\nModel Configuration Trials:")
print("  - Input chunk length: 168 (7 days)")
print("  - Output chunk length: 24 (1 day)")
print("  - Batch size: 4")
print("  - Epochs: 20")
print("  - Learning rate: 1e-5")
print("  - Quantiles: [0.1, 0.5, 0.9]")
```

```python
# Evaluation metrics
print("\nEvaluation Metrics:")
print("  - MAPE (Mean Absolute Percentage Error)")
print("  - MAE (Mean Absolute Error)")
print("  - RMSE (Root Mean Squared Error)")
print("  - Rolling rollout (6 chunks x 24h)")
```
